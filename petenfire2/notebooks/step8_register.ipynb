# Step 8 - Registrar modelo en MLflow Model Registry


```python
# Imports y configuracion del proyecto
from project_config import init_notebook
config = init_notebook()

import utils
logger = utils.init_logger('info')
```

```python
import mlflow
mlflow.set_tracking_uri(config['mlflow']['tracking_uri'])
client = mlflow.MlflowClient()

# Tomar el ultimo run del experimento
exp = client.get_experiment_by_name(config['mlflow']['experiment_name'])
runs = client.search_runs([exp.experiment_id], order_by=['attributes.start_time DESC'], max_results=1)
run = runs[0]
print('run_id:', run.info.run_id, 'val_pr:', run.data.metrics.get('val_pr_auc'))
```

## 8.1 Registrar modelo


```python
model_uri = f"runs:/{run.info.run_id}/stacking_ensemble.joblib"
result = mlflow.register_model(model_uri=model_uri, name=config['mlflow']['registry_name'])
print('registered version:', result.version)
```
